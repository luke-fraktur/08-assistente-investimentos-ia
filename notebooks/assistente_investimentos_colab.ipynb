# Assistente de Investimentos Educacional com IA

MVP demonstrativo para Google Colab, webhook e n8n. A resposta é educativa e não constitui recomendação financeira personalizada.

```python
!pip -q install flask pyngrok
from flask import Flask, request, jsonify

app = Flask(__name__)

def analisar(payload):
    pergunta = str(payload.get('question', '')).strip()
    if len(pergunta) < 10:
        raise ValueError('Informe uma pergunta com pelo menos 10 caracteres.')
    return {
        'type': 'educational_investment_briefing',
        'disclaimer': 'Material educativo; não constitui recomendação personalizada.',
        'question': pergunta,
        'asset': payload.get('asset', 'não informado'),
        'investor_profile': payload.get('investor_profile', 'não informado'),
        'horizon': payload.get('horizon', 'não informado'),
        'analysis_framework': ['fatos', 'hipóteses', 'riscos', 'dados ausentes'],
        'next_questions': ['Qual a fonte dos dados?', 'Quais riscos invalidam a hipótese?']
    }

@app.post('/analisar')
def analisar_endpoint():
    try:
        return jsonify(analisar(request.get_json(force=True) or {}))
    except ValueError as exc:
        return jsonify({'error': str(exc)}), 400

# Em uma célula final: app.run(host='0.0.0.0', port=5000)
```
